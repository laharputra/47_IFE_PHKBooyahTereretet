### 5. Kualitas data: duplikasi, pemeriksaan rentang, dan outlier
CSV ini memiliki **0 duplikat baris identik** dan **0 duplikat gabungan Date–Location**. IQR mendeteksi nilai ekstrem pada beberapa variabel, misalnya `Rainfall` **25.578 pengamatan** (17,58% seluruh baris). Karena hujan harian banyak bernilai 0, batas IQR bukan alasan menghapus data hujan ekstrem secara otomatis.

**Catatan okta:** Angka `9` pada laporan tutupan awan dapat berarti **langit tertutup fenomena lain seperti kabut** (bukan sekadar 9/8 tutupan awan), sehingga angka ini jangan langsung dianggap kesalahan. Lihat definisi okta pada sumber meteorologi dan validasi kembali kode yang digunakan oleh penyedia data.

````python
print("Duplikat seluruh kolom:", int(df.duplicated().sum()))
if {"Date", "Location"}.issubset(df.columns):
    print("Duplikat kombinasi Date & Location:", int(df.duplicated(subset=["Date", "Location"]).sum()))
checks = {}
for k in ["Humidity9am", "Humidity3pm"]:
    if k in df: checks[k + " di luar 0–100%"] = int(((df[k] < 0)|(df[k] > 100)).sum())
for k in ["Cloud9am", "Cloud3pm"]:
    if k in df: checks[k + " di luar 0–9 (9 dapat berarti langit tertutup fenomena lain)"] = int(((df[k] < 0)|(df[k] > 9)).sum())
for k in ["Rainfall", "Evaporation", "WindGustSpeed", "WindSpeed9am", "WindSpeed3pm"]:
    if k in df: checks[k + " bernilai negatif"] = int((df[k] < 0).sum())
print("Validasi sederhana rentang nilai:");display(pd.Series(checks, name="Jumlah_Nilai_Tidak_Valid").to_frame())
numeric_cols = df.select_dtypes(include="number").columns.drop("RISK_MM", errors="ignore")
outliers=[]
for k in numeric_cols:
    s = df[k].dropna()
    q1,q3 = s.quantile([.25,.75]);iqr=q3-q1
    n=int(((s < q1-1.5*iqr)|(s > q3+1.5*iqr)).sum())
    outliers.append((k,n,round(100*n/max(len(s),1),2)))
outliers_df=pd.DataFrame(outliers, columns=["Kolom", "Jumlah_Di_Luar_IQR", "Persen_NonMissing"]).sort_values("Jumlah_Di_Luar_IQR", ascending=False)
display(outliers_df)
````

### 4. Cleaning deterministik tanpa memanfaatkan informasi test

Aturan yang diterapkan:

1. Menghapus baris duplikat identik.
2. Memastikan kolom `Date` dapat dibaca sebagai tanggal; baris tanpa tanggal valid tidak dapat masuk pembagian kronologis.
3. Menghapus baris yang tidak memiliki label `RainTomorrow` valid (`Yes` atau `No`). **Label target tidak diimputasi.**
4. Nilai numerik yang jelas tidak valid (misalnya kelembapan >100%, curah hujan negatif) diubah menjadi `NaN` untuk ditangani melalui imputasi **berdasarkan training**.
5. Menghapus `RISK_MM` **jika ada**, karena mewakili informasi hujan yang berkaitan langsung dengan hari berikutnya dan berisiko membocorkan target.

**Catatan:** missing value pada fitur **belum diimputasi** di tahap ini. Imputasi dipasang dalam pipeline setelah train/test split.

````python
df_clean = df_raw.copy()
df_clean = df_clean.drop_duplicates().copy()
df_clean["Date"] = pd.to_datetime(df_clean["Date"], errors="coerce")
df_clean = df_clean.dropna(subset=["Date"]).copy()
df_clean = df_clean[df_clean["RainTomorrow"].isin(["Yes", "No"])].copy()

invalid_value_counts = {}
nonnegative_cols = ["Rainfall", "Evaporation", "Sunshine", "WindGustSpeed", "WindSpeed9am", "WindSpeed3pm"]
for col in nonnegative_cols:
    if col in df_clean.columns:
        bad = df_clean[col].notna() & (df_clean[col] < 0)
        invalid_value_counts[col] = int(bad.sum())
        df_clean.loc[bad, col] = np.nan

for col in ["Humidity9am", "Humidity3pm"]:
    if col in df_clean.columns:
        bad = df_clean[col].notna() & ~df_clean[col].between(0, 100)
        invalid_value_counts[col] = int(bad.sum())
        df_clean.loc[bad, col] = np.nan

# Kebocoran target pada beberapa varian dataset cuaca
leakage_cols = [c for c in ["RISK_MM"] if c in df_clean.columns]
df_clean = df_clean.drop(columns=leakage_cols)

comparison_cleaning = pd.DataFrame({
    "Indikator": ["Jumlah baris", "Missing target", "Duplikat baris", "Missing seluruh sel", "Missing pada fitur"],
    "Sebelum": [len(df_raw), int(df_raw["RainTomorrow"].isna().sum()),
               int(df_raw.duplicated().sum()), int(df_raw.isna().sum().sum()),
               int(df_raw.drop(columns=["RainTomorrow"]).isna().sum().sum())],
    "Sesudah cleaning awal": [len(df_clean), int(df_clean["RainTomorrow"].isna().sum()),
               int(df_clean.duplicated().sum()), int(df_clean.isna().sum().sum()),
               int(df_clean.drop(columns=["RainTomorrow"]).isna().sum().sum())],
})
display(comparison_cleaning)
print("Nilai tidak valid yang diubah menjadi NaN:", invalid_value_counts)
print("Fitur potensi leakage yang dibuang:", leakage_cols)
print("Catatan: fitur masih bisa mengandung missing, karena imputasi belum dilakukan.")
````

### 5. Train/test split kronologis (80% tanggal awal / 20% tanggal akhir)

Cutoff dihitung dari **80% tanggal unik pada dataset asli**, sesuai rancangan `02_Data_Understanding.ipynb`. Untuk CSV yang sama, cutoff yang diharapkan adalah **2015-08-08**.

Karena target `RainTomorrow` berarti hujan **hari berikutnya**, kita menerapkan **purge gap 1 hari kalender** pada akhir training. Pengamatan sehari sebelum cutoff tidak digunakan untuk training agar label training tidak memanfaatkan kejadian di tanggal awal testing. Ini membuat jumlah baris training sedikit lebih kecil daripada subset eksplorasi EDA sebelumnya.

**Larangan:** jangan menggunakan label atau distribusi data test untuk memutuskan metode imputasi, pemilihan fitur, atau hyperparameter.

````python
raw_dates = pd.to_datetime(df_raw["Date"], errors="coerce").dropna()
unique_dates = pd.Series(raw_dates.unique()).sort_values().reset_index(drop=True)
if len(unique_dates) < 10:
    raise ValueError("Tanggal unik terlalu sedikit untuk split kronologis")

cutoff = pd.Timestamp(unique_dates.iloc[int(len(unique_dates) * 0.80)])
purge_gap_days = 1
train_end_exclusive = cutoff - pd.Timedelta(days=purge_gap_days)

train_df = df_clean[df_clean["Date"] < train_end_exclusive].copy()
test_df = df_clean[df_clean["Date"] >= cutoff].copy()
purged_df = df_clean[(df_clean["Date"] >= train_end_exclusive) &
                     (df_clean["Date"] < cutoff)].copy()
if train_df.empty or test_df.empty:
    raise ValueError("Hasil split kosong. Periksa tanggal dan file CSV.")
assert train_df["Date"].max() < test_df["Date"].min()

print("Batas test:", cutoff.date())
print("Train sampai:", train_df["Date"].max().date())
print("Test mulai:", test_df["Date"].min().date())
print("Baris training:", len(train_df))
print("Baris test:", len(test_df))
print("Baris purge gap:", len(purged_df))
print("Baris label invalid yang sudah dibuang:", len(df_raw) - len(df_clean))

print("Distribusi target TRAIN:")
display(train_df["RainTomorrow"].value_counts(dropna=False).to_frame("Jumlah"))
print("Distribusi target TEST akan diperiksa pada bagian 7 setelah pilihan model final.")
````

### 6. Menyiapkan fitur dan target

- `RainTomorrow` adalah **target** dan tidak boleh masuk `X`.
- `Date` dipakai untuk split dan **dikeluarkan dari fitur awal**. Feature engineering terkait musim/tanggal dapat diputuskan pada tahap berikutnya.
- `Location` tetap menjadi fitur kategorikal.
- Semua kolom selain target, tanggal, dan fitur bocor menjadi **kandidat fitur**. Pemilihan fitur final dilakukan di notebook 04.

Nilai kosong pada fitur akan diproses dengan **median untuk fitur numerik** dan **modus untuk fitur kategorikal**. Fitur kategorikal menggunakan **One-Hot Encoding**. Fitur numerik menggunakan **StandardScaler** agar representasinya juga sesuai untuk Gaussian Naive Bayes; Random Forest sendiri umumnya tidak memerlukan scaling.

````python
excluded_features = ["RainTomorrow", "Date"]
feature_columns = [c for c in train_df.columns if c not in excluded_features]
X_train = train_df[feature_columns].copy()
X_test = test_df[feature_columns].copy()
y_train = train_df["RainTomorrow"].map({"No": 0, "Yes": 1}).astype(int)

numeric_features = X_train.select_dtypes(include="number").columns.tolist()
categorical_features = X_train.select_dtypes(exclude="number").columns.tolist()

print("Jumlah fitur kandidat:", len(feature_columns))
print("Fitur numerik (", len(numeric_features), "):", numeric_features)
print("Fitur kategorikal (", len(categorical_features), "):", categorical_features)
print("Dimensi X_train sebelum preprocessing:", X_train.shape)
print("Dimensi X_test sebelum preprocessing:", X_test.shape)
````

### 7. Pipeline preprocessing: fit pada TRAIN saja

**Pencegahan data leakage:** `.fit_transform()` dilakukan hanya pada `X_train`. Data `X_test` diproses memakai `.transform()` tanpa melakukan fit lagi. 

Pengubahan menjadi matriks *dense* dipilih agar nantinya kompatibel dengan `GaussianNB`. Jika memori perangkat terbatas, pendekatan sparse untuk Random Forest dan preprocessing khusus GaussianNB dapat dipisahkan pada tahap Modeling.

````python
numeric_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median", keep_empty_features=True)),
    ("scaler", StandardScaler()),
])
categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent", keep_empty_features=True)),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
])
preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features),
], remainder="drop", sparse_threshold=0)

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)
feature_names_after = preprocessor.get_feature_names_out()

print("X_train setelah preprocessing:", X_train_processed.shape)
print("X_test setelah preprocessing:", X_test_processed.shape)
print("Jumlah fitur hasil encoding:", len(feature_names_after))
print("Jumlah missing hasil train:", int(pd.isna(X_train_processed).sum()))
print("Jumlah missing hasil test:", int(pd.isna(X_test_processed).sum()))
assert np.isfinite(X_train_processed).all(), "Ada nilai NaN/inf dalam training setelah transformasi"
assert np.isfinite(X_test_processed).all(), "Ada nilai NaN/inf dalam testing setelah transformasi"

print("Contoh 15 fitur hasil encoding:")
display(pd.DataFrame({"Fitur": feature_names_after[:15]}))
````

Seluruh nilai hilang pada matriks fitur ditangani menggunakan parameter yang hanya dipelajari dari training.

**Interpretasi:** jumlah missing pada data mentah dan data hasil pembersihan awal mungkin berbeda karena ada baris dengan label target kosong yang dihapus. Missing pada dataset fitur setelah imputasi diharapkan 0, tanpa memakai statistik testing.

### 9. Ketidakseimbangan kelas dan strategi penanganan

### 3. Audit fitur **training**: missing value, tipe data, dan jumlah nilai unik

Fitur dengan missing tinggi berpotensi menambah ketidakpastian imputasi. Untuk **seleksi awal yang transparan**, kita mengusulkan membuang fitur dengan proporsi missing **≥ 40% pada training**. Angka 40% adalah **aturan praktis yang dipilih sebelum melihat hasil test**, bukan batas ilmiah universal; pengaruhnya perlu diuji pada validasi training.

Fitur berkorelasi rendah terhadap target **tidak langsung dibuang**, karena interaksi/nonlinearitas mungkin tetap berguna untuk Random Forest.

````python
feature_audit = pd.DataFrame({
    "Fitur": X_fit_selection.columns,
    "Tipe": X_fit_selection.dtypes.astype(str).values,
    "Missing TRAIN": X_fit_selection.isna().sum().values.astype(int),
    "Missing TRAIN (%)": (X_fit_selection.isna().mean() * 100).round(2).values,
    "Unik non-null": X_fit_selection.nunique(dropna=True).values.astype(int),
}).sort_values(["Missing TRAIN (%)", "Fitur"], ascending=[False, True]).reset_index(drop=True)
feature_audit["Calon drop: missing >= 40%"] = feature_audit["Missing TRAIN (%)"] >= MISSING_THRESHOLD * 100

display(feature_audit)

fig, ax = plt.subplots(figsize=(10, 6))
plot_data = feature_audit.sort_values("Missing TRAIN (%)", ascending=True)
ax.barh(plot_data["Fitur"], plot_data["Missing TRAIN (%)"])
ax.axvline(MISSING_THRESHOLD * 100, linestyle="--", label=f"Batas {MISSING_THRESHOLD:.0%}")
ax.set_xlabel("Persentase missing pada training (%)")
ax.set_title("Kelengkapan fitur pada TRAIN — tanpa melihat TEST")
ax.legend()
plt.tight_layout()
plt.show()
````

### 4. Hubungan fitur dengan target berdasarkan TRAIN: Mutual Information (MI)

**Mutual Information** mengukur besarnya informasi statistik fitur terhadap label dalam subset training (skor lebih besar = asosiasi terdeteksi lebih kuat, tetapi **bukan sebab-akibat**). Untuk efisiensi, evaluasi dilakukan pada sampel acak tetap, maksimal 20.000 baris training. Sampel **tidak berisi baris testing**.

- Fitur numerik: median diestimasi pada sampel training tersebut untuk pengukuran MI.
- Fitur kategorikal: modus dan kode kategori hanya untuk menghitung MI (bukan skema encoding final).
- Skor MI bukan ukuran kontribusi fitur pada model yang sudah dilatih; seleksi final akan diperiksa lewat validasi pada tahap Modeling.
- Hasil MI bisa bervariasi menurut sampling, terutama fitur dengan kategori banyak.

````python
mi_sample = X_fit_selection.sample(n=min(MI_SAMPLE_LIMIT, len(X_fit_selection)), random_state=RANDOM_STATE)
y_mi = y_fit_selection.loc[mi_sample.index]
num_cols = mi_sample.select_dtypes(include="number").columns.tolist()
cat_cols = [c for c in mi_sample.columns if c not in num_cols]

arrays = []
if num_cols:
    mi_num_imputer = SimpleImputer(strategy="median", keep_empty_features=True)
    arrays.append(mi_num_imputer.fit_transform(mi_sample[num_cols]).astype(float))
if cat_cols:
    mi_cat_imputer = SimpleImputer(strategy="most_frequent", keep_empty_features=True)
    cat_values = mi_cat_imputer.fit_transform(mi_sample[cat_cols]).astype(str)
    arrays.append(OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1).fit_transform(cat_values))
mi_matrix = np.column_stack(arrays)
mi_discrete = np.array([False] * len(num_cols) + [True] * len(cat_cols), dtype=bool)
mi_columns = num_cols + cat_cols
mi_scores = mutual_info_classif(mi_matrix, y_mi.to_numpy(), discrete_features=mi_discrete,
                                random_state=RANDOM_STATE, n_neighbors=3)
mi_table = pd.DataFrame({"Fitur": mi_columns, "MI (TRAIN saja)": mi_scores}).sort_values(
    "MI (TRAIN saja)", ascending=False).reset_index(drop=True)
print("Ukuran sampel training untuk MI:", len(mi_sample))
display(mi_table.round(5))

fig, ax = plt.subplots(figsize=(9, 5))
show_mi = mi_table.head(12).sort_values("MI (TRAIN saja)")
ax.barh(show_mi["Fitur"], show_mi["MI (TRAIN saja)"])
ax.set_xlabel("Mutual Information dengan RainTomorrow (TRAIN)")
ax.set_title("12 fitur berasosiasi tertinggi dalam sampel training")
plt.tight_layout()
plt.show()
````

### 5. Korelasi antarfitur numerik (TRAIN saja)

Korelasi Pearson dengan nilai absolut **≥ 0,95** menandai pasangan fitur yang **perlu ditinjau karena berpotensi redundan**. Ambang ini adalah aturan eksploratif, **bukan kewajiban menghapus salah satu fitur dari setiap pasangan**.

Dalam dataset yang dianalisis, terdapat dua pasangan menonjol (angka akhir mengikuti hasil eksekusi di bawah):

- **`Pressure9am` — `Pressure3pm`**: kedua variabel mengukur tekanan udara pada waktu berbeda. Untuk **baseline sederhana**, notebook mempertahankan `Pressure3pm` karena diukur lebih dekat ke hari berikutnya dan mengeluarkan `Pressure9am`. **Ini belum membuktikan `Pressure3pm` lebih informatif**. Justru perubahan tekanan pagi–sore berpotensi mengandung sinyal yang hilang jika `Pressure9am` dibuang.
- **`MaxTemp` — `Temp3pm`**: keduanya juga berkorelasi sangat tinggi. Namun, `MaxTemp` adalah **suhu maksimum harian**, sedangkan `Temp3pm` merupakan **suhu pada pengamatan sore**; informasi waktunya tidak persis sama. Untuk baseline, **keduanya tetap dipertahankan**, agar kemungkinan efek perbedaan waktu tidak dihilangkan sebelum divalidasi.

**Mengapa dua pasangan diperlakukan berbeda?** Ini adalah **keputusan rancangan baseline yang eksplisit, bukan aturan statistik otomatis**. Pasangan tekanan disederhanakan terlebih dahulu; pasangan suhu dipertahankan untuk eksperimen awal. Konsistensi dan manfaat dari kedua pilihan harus diuji dengan **validasi kronologis pada training saja**, misalnya membandingkan model dengan/ tanpa `Pressure9am`, dan dengan/ tanpa salah satu fitur suhu. **Testing tidak digunakan untuk mengambil keputusan ini.**


````python
numeric_train = X_fit_selection.select_dtypes(include="number")
corr = numeric_train.corr(numeric_only=True)
strong_pairs = []
for i, a in enumerate(corr.columns):
    for b in corr.columns[i+1:]:
        r = corr.at[a, b]
        if pd.notna(r) and abs(r) >= CORRELATION_THRESHOLD:
            strong_pairs.append({"Fitur A": a, "Fitur B": b, "Pearson r TRAIN": round(float(r), 4)})
strong_corr_table = pd.DataFrame(strong_pairs, columns=["Fitur A", "Fitur B", "Pearson r TRAIN"])
print("Pasangan dengan |r| >=", CORRELATION_THRESHOLD)
display(strong_corr_table if len(strong_corr_table) else pd.DataFrame({"Temuan":["Tidak ada pasangan yang mencapai ambang korelasi."]}))

fig, ax = plt.subplots(figsize=(9, 7))
image = ax.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
ax.set_xticks(np.arange(len(corr.columns)), labels=corr.columns, rotation=80, fontsize=8)
ax.set_yticks(np.arange(len(corr.columns)), labels=corr.columns, fontsize=8)
ax.set_title("Korelasi Pearson fitur numerik — TRAIN")
fig.colorbar(image, ax=ax, shrink=.75)
plt.tight_layout()
plt.show()
````

### 6. Keputusan Feature Selection: fitur **dipakai** dan **dibuang**

Aturan baseline awal yang mudah diaudit untuk Progress 2:

1. **Missing value tinggi:** buang fitur yang memiliki missing **≥ 40% pada training**. Ambang 40% merupakan **heuristik**, bukan batas ilmiah universal. Khusus `Sunshine`, skor *Mutual Information* (MI) pada sampel training relatif tinggi; artinya **fitur ini mungkin berguna**, meskipun diputuskan untuk dibuang karena banyak nilai kosong. Eksperimen Modeling perlu membandingkan hasil **dengan dan tanpa `Sunshine`** menggunakan metode imputasi yang sesuai.
2. **Redundansi tekanan:** jika korelasi `Pressure9am`–`Pressure3pm` pada training mencapai `|r| ≥ 0,95`, baseline **membuang `Pressure9am` dan mempertahankan `Pressure3pm`**. Alasan: penyederhanaan kandidat dan mempertahankan pengamatan sore yang lebih dekat ke hari berikutnya. **Alasan ini bukan bukti bahwa `Pressure9am` kurang prediktif.** Fitur tekanan pagi dan perubahan tekanan antarwaktu harus ditinjau kembali lewat validasi training.
3. **Korelasi suhu:** meskipun `MaxTemp` dan `Temp3pm` juga berkorelasi sangat tinggi, **keduanya tetap dipakai** dalam baseline karena mewakili ringkasan suhu maksimum dan pengamatan suhu pukul 15.00 yang tidak identik secara makna. Bandingkan alternatif penghapusan salah satu fitur pada tahap Modeling; **jangan menyatakan keduanya pasti diperlukan sebelum ada hasil validasi**.
4. **Variasi dan keamanan target:** buang fitur yang konstan pada training. `Date` dipakai hanya untuk split kronologis; `RainTomorrow` adalah label dan tidak boleh masuk `X`. `RISK_MM` (jika tersedia) dibuang karena berpotensi membocorkan target.
5. **Fitur lainnya:** pertahankan sebagai kandidat baseline, termasuk `Location` dan `RainToday`, walaupun MI rendah. MI tidak mengukur kontribusi setiap fitur pada model yang sudah dilatih.

**Penting:** keputusan **PAKAI/BUANG** di bawah merupakan **baseline pra-model**, bukan hasil optimasi performa. Semua aturan dan statistik seleksi dihitung pada **training**, tanpa `y_test` dan tanpa statistik fitur testing.


````python
missing_ratio_train = X_fit_selection.isna().mean()
remove_missing = {c for c in X_fit_selection.columns if missing_ratio_train[c] >= MISSING_THRESHOLD}
remove_constant = {c for c in X_fit_selection.columns if X_fit_selection[c].nunique(dropna=True) <= 1}
remove_redundant = set()

if {"Pressure9am", "Pressure3pm"}.issubset(X_fit_selection.columns):
    pressure_r = corr.loc["Pressure9am", "Pressure3pm"]
    if pd.notna(pressure_r) and abs(pressure_r) >= CORRELATION_THRESHOLD:
        remove_redundant.add("Pressure9am")

drop_features = remove_missing | remove_constant | remove_redundant
selected_features = [c for c in feature_columns if c not in drop_features]
if not selected_features:
    raise ValueError("Tidak ada fitur tersisa. Tinjau aturan seleksi.")

mi_lookup = mi_table.set_index("Fitur")["MI (TRAIN saja)"].to_dict()
decisions = []
for c in feature_columns:
    reasons = []
    if c in remove_missing: reasons.append(f"Missing TRAIN >= {MISSING_THRESHOLD:.0%}")
    if c in remove_constant: reasons.append("Konstan/tanpa variasi di TRAIN")
    if c in remove_redundant:
        reasons.append("Penyederhanaan baseline pasangan tekanan; validasi ulang Pressure9am vs Pressure3pm")
    if not reasons:
        if c == "MaxTemp":
            reasons.append("Dipertahankan: suhu maksimum harian; uji redundansi dengan Temp3pm saat Modeling")
        elif c == "Temp3pm":
            reasons.append("Dipertahankan: suhu pengamatan sore; uji redundansi dengan MaxTemp saat Modeling")
        elif c == "Pressure3pm":
            reasons.append("Dipertahankan: tekanan sore untuk baseline; bandingkan dengan opsi memakai Pressure9am")
        else:
            reasons.append("Dipertahankan untuk baseline; validasi manfaat saat Modeling")
    if c in remove_missing and c == "Sunshine":
        reasons.append("MI training relatif tinggi; bandingkan dengan opsi mempertahankan Sunshine")
    decisions.append({
        "Fitur asli": c,
        "Missing TRAIN (%)": round(100*float(missing_ratio_train[c]),2),
        "MI TRAIN (sampel)": round(float(mi_lookup.get(c,0)),5),
        "Keputusan": "BUANG" if c in drop_features else "PAKAI",
        "Alasan": "; ".join(reasons)
    })
decision_table = pd.DataFrame(decisions).sort_values(["Keputusan", "Fitur asli"]).reset_index(drop=True)
display(decision_table)
print("Fitur dibuang (", len(drop_features), "):", sorted(drop_features))
print("Fitur dipertahankan (", len(selected_features), "):", selected_features)
assert all(c in X_fit_selection.columns for c in selected_features)
assert "RainTomorrow" not in selected_features and "Date" not in selected_features
````

### Pemeriksaan konsistensi keputusan pada pasangan berkorelasi tinggi

Tabel berikut menampilkan **kedua pasangan fitur** dan status **PAKAI/BUANG** menurut aturan baseline. Dengan cara ini, keputusan mempertahankan dua fitur suhu tetapi membuang satu fitur tekanan menjadi **terbuka dan dapat dikritisi** pada presentasi.

Keputusan ini **bukan hasil uji performa**. Pada tahap 05, lakukan pembandingan *ablation* dan/atau validasi waktu **hanya di training** sebelum menetapkan fitur final.

````python
if not strong_corr_table.empty:
    correlated_decisions = strong_corr_table.copy()
    correlated_decisions["Keputusan A"] = correlated_decisions["Fitur A"].map(
        lambda name: "BUANG" if name in drop_features else "PAKAI")
    correlated_decisions["Keputusan B"] = correlated_decisions["Fitur B"].map(
        lambda name: "BUANG" if name in drop_features else "PAKAI")
    def rationale(row):
        pair = {row["Fitur A"], row["Fitur B"]}
        if pair == {"Pressure9am", "Pressure3pm"}:
            return "Sederhanakan baseline tekanan; uji lagi fitur pagi/perubahan tekanan"
        if pair == {"MaxTemp", "Temp3pm"}:
            return "Pertahankan dua informasi suhu berbeda; uji ablation saat Modeling"
        return "Pasangan lain: pertahankan/seleksi menurut aturan baseline dan validasi training"
    correlated_decisions["Alasan keputusan baseline"] = correlated_decisions.apply(rationale, axis=1)
    display(correlated_decisions)
else:
    print("Tidak ada pasangan fitur numerik yang melewati ambang korelasi.")
````

### 7. Terapkan seleksi ke training dan testing — **fit preprocessing pada training saja**

Seleksi dibuat **hanya berdasarkan training** di bagian 3–6; di sini testing baru dipakai untuk transformasi dengan kolom yang sama. Kode berikut mengulang strategi pipeline notebook 03: median numerik, modus kategorikal, one-hot encoding, StandardScaler numerik.

> **Peringatan metodologis:** meskipun ada tahap `fit` pada objek preprocessing, kode ini **tidak melatih model prediksi**. Data test tidak di-*fit*, hanya ditransformasi.

````python
X_train_selected = X_fit_selection[selected_features].copy()
X_test_selected = valid_df[selected_features].copy()

selected_num = X_train_selected.select_dtypes(include="number").columns.tolist()
selected_cat = [c for c in selected_features if c not in selected_num]

transformers = []
if selected_num:
    transformers.append(("num", Pipeline([
        ("imputer", SimpleImputer(strategy="median", keep_empty_features=True)),
        ("scaler", StandardScaler())]), selected_num))
if selected_cat:
    transformers.append(("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent", keep_empty_features=True)),
        ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))]), selected_cat))
preprocessor_selected = ColumnTransformer(transformers, remainder="drop", sparse_threshold=0)
X_train_selected_processed = preprocessor_selected.fit_transform(X_train_selected)
X_test_selected_processed = preprocessor_selected.transform(X_test_selected)
selected_encoded_names = preprocessor_selected.get_feature_names_out()

assert X_train_selected_processed.shape[0] == len(y_fit)
assert X_test_selected_processed.shape[0] == len(valid_df)
assert X_train_selected_processed.shape[1] == X_test_selected_processed.shape[1]
assert np.isfinite(X_train_selected_processed).all() and np.isfinite(X_test_selected_processed).all()

comparison = pd.DataFrame({
    "Indikator": ["Jumlah baris fit internal", "Jumlah baris validation", "Fitur asli untuk modeling",
                  "Fitur dibuang", "Kolom numerik terpilih", "Kolom kategorikal terpilih",
                  "Kolom matriks hasil encoding", "Missing/inf setelah preprocessing"],
    "Hasil": [len(X_train_selected),len(X_test_selected),len(selected_features),len(drop_features),
              len(selected_num),len(selected_cat),len(selected_encoded_names),
              int((~np.isfinite(X_train_selected_processed)).sum() + (~np.isfinite(X_test_selected_processed)).sum())]
})
display(comparison)
print("Bentuk akhir FIT:", X_train_selected_processed.shape)
print("Bentuk akhir VALIDATION:", X_test_selected_processed.shape)
print("Nama 15 kolom pertama hasil transformasi:")
display(pd.DataFrame({"Fitur": selected_encoded_names[:15]}))
````

### 8. Simpan daftar fitur sebagai konfigurasi (bukan menyimpan data training/test)

File `feature_selection_config.json` menyimpan daftar fitur dan aturan seleksi, sehingga proses Modeling dapat memakai keputusan yang sama. ### 6. Membangun pipeline preprocessing dan dua model

Semua langkah *imputation*, *scaling*, dan *one-hot encoding* berada di dalam `Pipeline`. Dengan demikian, nilai untuk mengisi missing, rata-rata/standar deviasi, serta kategori encoding dipelajari dari **bagian fit saja**; validation hanya menerima transformasi.

- **Gaussian Naive Bayes** memakai fitur numerik yang sudah ditransformasi.
- **Random Forest** menggunakan 100 pohon, kedalaman maksimum 12, dan bobot kelas seimbang sebagai konfigurasi awal agar waktu eksekusi wajar.

Ini adalah konfigurasi eksperimen awal, **bukan klaim hyperparameter yang pasti optimal**.


````python
MODEL_PARAMETERS = {
    "Gaussian Naive Bayes": {"var_smoothing": 1e-9},
    "Random Forest": {
        "n_estimators": 100,
        "max_depth": 12,
        "min_samples_leaf": 3,
        "class_weight": "balanced",
        "random_state": RANDOM_STATE,
        "n_jobs": -1,
    },
}

def make_preprocessor(columns):
    numeric = fit_df[columns].select_dtypes(include="number").columns.tolist()
    categorical = [c for c in columns if c not in numeric]
    steps = []
    if numeric:
        steps.append(("num", Pipeline([
            ("imputer", SimpleImputer(strategy="median", keep_empty_features=True)),
            ("scaler", StandardScaler()),
        ]), numeric))
    if categorical:
        steps.append(("cat", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent", keep_empty_features=True)),
            ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
        ]), categorical))
    return ColumnTransformer(steps, remainder="drop", sparse_threshold=0)

def make_pipeline(name, columns):
    if name == "Gaussian Naive Bayes":
        estimator = GaussianNB(**MODEL_PARAMETERS[name])
    elif name == "Random Forest":
        estimator = RandomForestClassifier(**MODEL_PARAMETERS[name])
    else:
        raise ValueError(f"Nama model tidak dikenal: {name}")
    return Pipeline([("preprocessing", make_preprocessor(columns)), ("model", estimator)])

print("Model yang digunakan:", list(MODEL_PARAMETERS))
````

### 7. Training dan evaluasi pada validation

Tiap pasangan model dan varian fitur dilatih **terpisah**, kemudian diuji pada validation yang sama. Kelas positif adalah `Yes` (1).

Metrik:

- **Accuracy:** proporsi prediksi benar secara keseluruhan.
- **Precision (Yes):** ketepatan saat model memprediksi hujan.
- **Recall (Yes):** proporsi kejadian hujan yang berhasil terdeteksi.
- **F1-score (Yes):** keseimbangan precision dan recall untuk kelas hujan.

*Catatan:* Training seluruh kombinasi dapat memakan beberapa menit, terutama Random Forest.


````python
validation_records = []
validation_predictions = {}

for variant_name, columns in variants.items():
    X_fit_part = fit_df[columns]
    X_valid_part = valid_df[columns]
    for model_name in MODEL_PARAMETERS:
        print(f"Melatih {model_name} | fitur: {variant_name} ({len(columns)} fitur)...")
        pipeline = make_pipeline(model_name, columns)
        pipeline.fit(X_fit_part, y_fit)
        predictions = pipeline.predict(X_valid_part)

        key = (variant_name, model_name)
        validation_predictions[key] = predictions
        validation_records.append({
            "varian": variant_name,
            "model": model_name,
            "jumlah_fitur": len(columns),
            "accuracy": float(accuracy_score(y_valid, predictions)),
            "precision_yes": float(precision_score(y_valid, predictions, pos_label=1, zero_division=0)),
            "recall_yes": float(recall_score(y_valid, predictions, pos_label=1, zero_division=0)),
            "f1_yes": float(f1_score(y_valid, predictions, pos_label=1, zero_division=0)),
        })

results_df = pd.DataFrame(validation_records).sort_values(
    ["f1_yes", "recall_yes", "precision_yes"], ascending=False
).reset_index(drop=True)
display(results_df.round(4))
````

# 7. Model Evaluation — Data Testing

Data testing merupakan periode historis setelah cutoff dan **tidak dipakai untuk memilih fitur, model, atau hyperparameter**. Untuk memenuhi tujuan proyek dua algoritma, kita menilai **masing-masing model sekali** pada test dengan varian fitur terbaiknya berdasarkan validation. Model utama tetap dipilih menggunakan validation, bukan setelah membaca hasil test.

Metrik utama: accuracy, precision, recall, dan F1-score untuk kelas hujan (`Yes`). Confusion matrix membantu mengidentifikasi *false negative* (hari hujan diprediksi tidak hujan).


````python
from sklearn.metrics import classification_report

y_test_final = test_df["RainTomorrow"].map({"No": 0, "Yes": 1}).astype(int)
print("Distribusi target TEST (baru diperiksa pada tahap evaluasi akhir):")
display(y_test_final.value_counts().rename(index={0: "No", 1: "Yes"}).to_frame("Jumlah"))
selected_by_algorithm = results_df.sort_values(
    ["f1_yes", "recall_yes", "precision_yes"], ascending=False
).drop_duplicates(subset=["model"], keep="first")

test_models = {}
test_predictions = {}
test_probabilities = {}
test_scores = []

for _, record in selected_by_algorithm.iterrows():
    name = str(record["model"])
    variant = str(record["varian"])
    feature_list = list(variants[variant])

    # Kandidat pemenang telah dilatih ulang pada akhir bagian 6.
    if name == best_model_name and variant == best_variant:
        model = final_pipeline
    else:
        model = make_pipeline(name, feature_list)
        model.fit(train_pool[feature_list], y_train_full)

    # Test hanya diprediksi satu kali per kandidat, tanpa penyesuaian lanjutan.
    pred = model.predict(test_df[feature_list])
    proba = model.predict_proba(test_df[feature_list])[:, 1]
    test_models[name] = {"pipeline": model, "variant": variant, "features": feature_list}
    test_predictions[name] = pred
    test_probabilities[name] = proba
    test_scores.append({
        "Model": name,
        "Varian validasi": variant,
        "Jumlah fitur": len(feature_list),
        "Accuracy": float(accuracy_score(y_test_final, pred)),
        "Precision Yes": float(precision_score(y_test_final, pred, pos_label=1, zero_division=0)),
        "Recall Yes": float(recall_score(y_test_final, pred, pos_label=1, zero_division=0)),
        "F1 Yes": float(f1_score(y_test_final, pred, pos_label=1, zero_division=0)),
    })

test_results_df = pd.DataFrame(test_scores).sort_values("F1 Yes", ascending=False).reset_index(drop=True)
print("EVALUASI FINAL PADA TEST (tidak digunakan untuk memilih model):")
display(test_results_df.round(4))
print("Baris test:", len(y_test_final), "| Positif aktual:", int(y_test_final.sum()))
````
